```python
#9. Classes and Object-Oriented Programming

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

    def greet(self):
        print(f"Hello, my name is {self.name}")

p1 = Person("Alice", 25)
p1.greet()
      
```

```python
#10. Built-in Methods and Functions

# String methods
text = "hello"
print(text.upper())

# List methods
numbers = [1, 2, 3]
numbers.append(4)
print(numbers)
      
```

```python
#11. List Comprehensions

squares = [x**2 for x in range(5)]
print(squares)
      
```

```python
#12. Decorators

def decorator_func(func):
    def wrapper():
        print("Before function call")
        func()
        print("After function call")
    return wrapper

@decorator_func
def say_hello():
    print("Hello!")

say_hello()
      
```

```python
#13. Generators and Iterators

def gen():
    for i in range(3):
        yield i

for value in gen():
    print(value)
      
```

```python
#14. Regular Expressions

import re
pattern = r'\d+'
text = "There are 24 hours in a day"
match = re.findall(pattern, text)
print(match)
      
```
